```python
import toytree
import toyplot
```

## IQ-TREE (maximum likelihood)

```python
# run this on command line

# iqtree2 -s panand_FINAL_nov2025.fasta -m TVM+F+R10 -bb 1000 -nt 16 -seed 12345
```

```python
newick = '../phylogenies/panand_FINAL_nov2025.fasta.contree'
panand = toytree.tree(newick, internal_labels='support')
panand = panand.root('Paspalum_vaginatum')
ladder = panand.ladderize()
```

```python
canvas, axes, mark = ladder.draw(tip_labels_align=True,
                                 tree_style='n',
                                 node_labels='support',
                                 node_sizes=16,
                                 node_colors='white',
                                 height = 15000,
                                 width = 3000,
                                 use_edge_lengths = True);
```

```python
toytree.save(canvas, "../phylogenies/panand_iqtree_FINAL.pdf")
```
